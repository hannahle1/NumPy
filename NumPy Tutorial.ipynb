```python
# Repeat an array
arr = np.array([[1,2,3]]) # repeat the inner part
r1 = np.repeat(arr,3, axis=0) # axis = 0 -> downwards, across rows
print(r1)
```

```python
arr = np.array([[1,2,3]])
r2 = np.repeat(arr, 3, axis=1) # axis = 1 -> from left to right through columns
print(r2)
```

  
The solution is below

**1D Matrix** (Vector/ Array): 1 axis (length N) -> A[i]  
**2D Matrix** (Grid/ Table): 2 axes (rows R and columns C) -> A[i][j]  
**3D Matrix** (Cube/ Tensor): 3 axes (DxRxC) -> A[i][j][k] -> A stack of pages (a book)   
**4D Matrix** (Hypercube/ Batch Tensor): 4 axes (NxDxHxW) -> A[i][j][k][l] -> A collection of books/ video frames

```python
output = np.ones((5,5))
print(output)

z = np.zeros((3,3))
z[1,1] = 9
print(z)
# uses 2D slicing to overwrite the inner region of output with z.
# 1: -1 (select rows starting at index 1 -> rows 1,2,3)
# 1: -1 (select columns starting at index 1 -> columns 1,2,3)
output[1:-1,1:-1] = z 
print(output)
```

## Linear Algebra

```python
a = np.ones((2,3)) # (2,4) will not work because it's not compatible. (2,4)x(3,2) will not work
print(a)

b = np.full((3,2), 2)
print(b)

np.matmul(a,b) # a*b will not work because they're different sizes
```

```python
# Find the determinant
c = np.identity(3) # identity matrix
np.linalg.det(c) # linalg: linear algebra
```

## Reorganizing Arrays

```python
before = np.array([[1,2,3,4],[5,6,7,8]])
print(before)

after = before.reshape((8,1)) # if (2,3) -> error because 2x3=6 elements != 8 elements
print(after)
```

```python
# Vertically stacking vectors
v1 = np.array([1,2,3,4])
v2 = np.array([5,6,7,8])

np.vstack([v1,v2,v1,v2]) # if the sizes are mismatched -> can't stack
```

```python
# Horizontal  stack
h1 = np.ones((2,4))
h2 = np.zeros((2,2))

np.hstack((h1,h2))
```
